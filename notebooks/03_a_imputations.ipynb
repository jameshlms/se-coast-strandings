```python
from prophet import Prophet
import pandas as pd
```

```python
df = pd.read_parquet("../data/processed/copepod_dataset_se_coast.parquet")
```

```python
df.head()
```

```python
temp_df = df[["date", "count"]].rename(columns={"date": "ds", "count": "y"})
temp_df["y"] = (
    pd.to_numeric(temp_df["y"], errors="coerce").dropna().reset_index(drop=True)
)
```

```python
model = Prophet()
model.fit(temp_df)
future = model.make_future_dataframe(periods=365)
forecast = model.predict(future)
```
